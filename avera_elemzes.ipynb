```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Globális Tágulás (Logfile.dat) beolvasása
print("Logfile betöltése...")
log_data = np.loadtxt('LCDM/Logfile.dat')
time = log_data[:, 0]
hubble = log_data[:, 4]

# 2. Kozmikus Háló (t1400.dat) beolvasása
print("Snapshot betöltése (ez eltarthat egy kis ideig)...")
snap_data = np.loadtxt('LCDM/t1400.dat')
x = snap_data[:, 0]
y = snap_data[:, 1]
z = snap_data[:, 2]

# Vágjunk ki egy 20 Mpc vastag "szeletet" a dobozból, hogy lássuk a struktúrát
slice_mask = (z > 70) & (z < 90)
x_slice = x[slice_mask]
y_slice = y[slice_mask]

# Rajzolás beállítása
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Bal ábra: Tágulástörténet
ax1.plot(time, hubble, color='blue', linewidth=2, label='H(t)')
ax1.set_title("Globális Tágulási Ütem (H) az idő függvényében")
ax1.set_xlabel("Szimulációs Idő")
ax1.set_ylabel("Hubble paraméter")
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.legend()

# Jobb ábra: Kozmikus Háló metszet
ax2.scatter(x_slice, y_slice, s=0.5, color='black', alpha=0.6)
ax2.set_title("Kozmikus Háló Metszet a jelenben (t1400)")
ax2.set_xlabel("X koordináta (Mpc/h)")
ax2.set_ylabel("Y koordináta (Mpc/h)")
ax2.set_aspect('equal') # Hogy a tér négyzetes maradjon

plt.tight_layout()
plt.show()
```
